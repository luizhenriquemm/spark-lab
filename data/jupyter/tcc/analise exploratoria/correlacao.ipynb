```python
cols_numericas = [
    'uti_int_to',
    'qt_proc',
    'idade',
    'dias_perm',
    'feriados_qtd',
    'feriados_prox_qtd',
    'feriados_prox_dist',
    'munic_res_populacao_residente',
    'munic_mov_populacao_residente',
    'munic_res_area_unidade_territorial',
    'munic_mov_area_unidade_territorial',
    'munic_res_idhm',
    'munic_mov_idhm'
]

cols_categoricas = [
    'dt_atend_dia',
    'dt_atend_ano',
    'dt_atend_woy',
    'dt_nasc_dia',
    'dt_nasc_ano',
    'dt_nasc_woy',
    'dt_saida_dia',
    'dt_saida_ano',
    'dt_saida_woy',
    'sexo',
    'dt_atend_dow',
    'feriado',
    'diag_princ_capitulo_numero',
    'proc_rea',
    'dt_atend_mes',
    'dt_nasc_mes',
    'dt_saida_mes',
    'diag_princ',
    'munic_res_sigla_uf',
    'munic_mov_sigla_uf',
    'munic_res_nome_uf',
    'munic_mov_nome_uf',
    'diag_princ_categoria',
    'morte'
]

target_col = 'morte'
```

```python
# Categórica ↔ Categórica → Cramér’s V ou Theil’s U
# Categórica ↔ Numérica → Eta² (ANOVA)
# Numérica ↔ Numérica → Pearson / Spearman / Kendall

# Pearson: assume que os números têm relação linear e métrica contínua. Para categóricas não é adequado, porque os números que você atribuiu não necessariamente têm relação matemática.
# Spearman: mede correlação de ordem (rank).
#     Funciona bem para variáveis ordinais (ex.: 1 = baixo, 2 = médio, 3 = alto).
#     Não é indicado para variáveis nominais (ex.: 1 = azul, 2 = verde, 3 = vermelho), porque não existe uma ordem natural.
```

```python
corr_done = []
```

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import Correlation
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import ChiSquareTest

skip = [
    # "diag_princ"
]

n = df_prep_corrigido.count()
def corr_cat_cat(col_1, col_2): # ChiSquareTest
    assembler = VectorAssembler(inputCols=[col_1], outputCol="features")
    df_features = assembler.transform(df_prep_corrigido.select(col_1, col_2))
    r = ChiSquareTest.test(df_features, "features", col_2).head()
    cramers_v = []
    for i, chi2 in enumerate(r.statistics):
        k = b.min(df_prep_corrigido.select([col_1, col_2][i]).distinct().count(),
                df_prep_corrigido.select(col_2).distinct().count())
        v = np.sqrt(chi2 / (n * (k-1)))
        cramers_v.append(([col_1, col_2][i], v))
    return cramers_v[0][1]

def corr_cat_num(num_col, cat_col): # eta squared
    global_mean = df_prep_corrigido.agg(mean(num_col)).first()[0]
    ss_total = df_prep_corrigido.withColumn("sq_diff", (col(num_col) - global_mean) ** 2) \
                 .agg(sum("sq_diff")).first()[0]

    group_stats = df_prep_corrigido.groupBy(cat_col) \
                    .agg(
                        count(num_col).alias("n"),
                        mean(num_col).alias("mean")
                    )

    ss_between = group_stats.withColumn(
        "sq_between",
        col("n") * (col("mean") - global_mean) ** 2
    ).agg(sum("sq_between")).first()[0]

    eta2 = ss_between / ss_total if ss_total != 0 else 0.0
    return eta2

def corr_num_num(col_1, col_2): # spearman
    assembler = VectorAssembler(inputCols=[col_1, col_2], outputCol="features")
    df_vector = assembler.transform(df_prep_corrigido)
    correlation_matrix = Correlation.corr(df_vector, "features", "spearman").head()[0]
    correlation_array = np.array(correlation_matrix.toArray())
    return correlation_array[0, 1]

for col_n in cols_numericas:
    for col_n2 in cols_numericas:
        is_done = False
        for i in corr_done:
            if col_n in i["cols"] and col_n2 in i["cols"]:
                is_done = True
        if col_n == col_n2:
            is_done = True
        if not is_done:
            print(f"Doing for {col_n} x {col_n2} ")
            corr_done.append({
                "cols": [col_n, col_n2],
                "method": "spearman",
                "result": corr_num_num(col_n, col_n2)
            })
            
    for col_c in cols_categoricas:
        is_done = False
        for i in corr_done:
            if col_n in i["cols"] and col_c in i["cols"]:
                is_done = True
        if not is_done:
            print(f"Doing for {col_n} x {col_c} ")
            corr_done.append({
                "cols": [col_n, col_c],
                "method": "eta_squared",
                "result": corr_cat_num(col_n, col_c)
            })

for col_c in cols_categoricas:
    for col_c2 in cols_categoricas:
        is_done = False
        for i in corr_done:
            if col_c in i["cols"] and col_c2 in i["cols"]:
                is_done = True
        if col_c == col_c2 or col_c in skip: #  or col_c2 in skip
            is_done = True
        if not is_done:
            print(f"Doing for {col_c} x {col_c2} ")
            try:
                corr_done.append({
                    "cols": [col_c, col_c2],
                    "method": "cramers_v",
                    "result": corr_cat_cat(col_c, col_c2)
                })
            except:
                corr_done.append({
                    "cols": [col_c, col_c2],
                    "method": "cramers_v",
                    "result": 0.0
                })
```

```python
# Construir matriz simétrica
matrix = df.pivot(index="col1", columns="col2", values="result")

# Tornar simétrica (preencher o outro lado)
matrix = matrix.combine_first(matrix.T)

# Preencher diagonal com 1
for c in matrix.columns:
    matrix.loc[c, c] = 1.0

matrix = matrix.fillna(0)
    
# print(matrix)
```
